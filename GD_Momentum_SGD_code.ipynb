## 1. Gradient Descent with Momentum



```python
import numpy as np

# check convergence
def has_converged(theta_new, grad):
    return np.linalg.norm(grad(theta_new)) / len(theta_new) < 1e-3


def GD_momentum(theta_init, grad, eta, gamma):
    # Suppose we want to store history of theta
    theta = [theta_init]

    # Khởi tạo vận tốc ban đầu bằng 0
    v_old = np.zeros_like(theta_init)

    for it in range(100):
        # Cập nhật vận tốc
        v_new = gamma * v_old + eta * grad(theta[-1])

        # Cập nhật theta
        theta_new = theta[-1] - v_new

        # Kiểm tra hội tụ
        if has_converged(theta_new, grad):
            break

        # Lưu theta mới
        theta.append(theta_new)

        # Lưu vận tốc để dùng cho vòng lặp tiếp theo
        v_old = v_new

    return theta

# theta chứa toàn bộ các điểm trên đường đi.
# Nếu chỉ muốn lấy nghiệm cuối cùng:
# return theta[-1]
```
